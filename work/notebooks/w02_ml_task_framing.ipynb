## My lane as an ML task

This is a classification problem.

The goal is to predict whether a content page needs to be refreshed based on its performance and content features. The output has two classes: Needs Refresh and Does Not Need Refresh.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*


## Target or Proxy

The dataset does not contain a direct target called needs_refresh.

I use the trend_direction column as a proxy target.

If trend_direction is "down", the page is labeled as needing refresh. Otherwise, it is labeled as not needing refresh.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

df["needs_refresh"] = (df["trend_direction"]=="down").astype(int)

df[["trend_direction","needs_refresh"]].head()
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*


## Success Metric

The success metric is F1-score.

F1-score balances precision and recall and is appropriate for this binary classification problem.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

print("Success Metric: F1-score")

df["needs_refresh"].value_counts()
```
